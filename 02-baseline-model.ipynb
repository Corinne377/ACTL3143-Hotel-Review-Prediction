# 02 — Baseline Model: Ridge Regression
Fits Ridge on TF-IDF + numeric features. Metrics computed in 05-analysis.ipynb.
> **Runtime: <1 min**

```python
# ── All imports ──────────────────────────────────────────────────────────
import numpy as np
import pickle
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error, mean_absolute_error
```

```python
# ── Load data ─────────────────────────────────────────────────────────────
X_train = np.load("data/X_train_dense.npy")
X_val   = np.load("data/X_val_dense.npy")
y_train = np.load("data/y_train.npy")
y_val   = np.load("data/y_val.npy")
print(f"Train: {X_train.shape[0]:,} × {X_train.shape[1]:,}")
```

```python
# ── Fit Ridge (α=10) ──────────────────────────────────────────────────────
print("Fitting Ridge Regression (alpha=10)...")
ridge = Ridge(alpha=10.0)
ridge.fit(X_train, y_train)

val_preds = ridge.predict(X_val)
print(f"Val MAE : {mean_absolute_error(y_val, val_preds):.4f}")
print(f"Val RMSE: {mean_squared_error(y_val, val_preds)**0.5:.4f}")
```

```python
with open("data/ridge_model.pkl","wb") as f: pickle.dump(ridge, f)
print("✅ Ridge model saved to data/ridge_model.pkl")
```
